```python
from langchain_core.messages import HumanMessage
from langchain_ollama import ChatOllama
from pydantic import BaseModel, Field
import pandas as pd

# 1. Inherit from BaseModel and use proper type hinting syntax (:)
class Dataset(BaseModel):
    columns: list[str] = Field(description="Name of the columns for dataset generator")
    data: list[list] = Field(description="Data/rows for creating dataset")

# 2. Initialize the model
model = ChatOllama(model="mistral")
structured_model = model.with_structured_output(Dataset)

# 3. Use a clean string or HumanMessage for the prompt
prompt = "Generate 5 sample product details for mobile phones. Give me columns like product_id, brand, model, price, and ram."

# 4. Invoke the structured model
result = structured_model.invoke(prompt)

# 5. Convert the structured response directly into a pandas DataFrame
df = pd.DataFrame(result.data, columns=result.columns)
print(df)
```
